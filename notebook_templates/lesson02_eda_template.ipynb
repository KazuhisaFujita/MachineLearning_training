```python
# ===== Step 0: ライブラリ読み込み =====
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

sns.set_theme(style='whitegrid')
```

```python
# ===== Step 1: データ読み込み =====
# TODO: IrisデータをDataFrameとして準備する
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

display(df.head())
```

```python
# ===== Step 2: pairplot =====
# TODO: hue に使う列を確認する
pairplot_columns = iris.feature_names + ['species']
sns.pairplot(df[pairplot_columns], hue='species', corner=True)
plt.show()
```

```python
# ===== Step 3: 箱ひげ図 =====
# TODO: どの特徴量が品種ごとに分かれやすいか観察する
target_feature = 'petal length (cm)'  # TODO: 別の列にも変更して試す

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='species', y=target_feature)
plt.title(f'{target_feature} の箱ひげ図')
plt.show()
```

```python
# ===== Step 4: 相関行列 =====
# TODO: 数値列だけを使って相関行列を確認する
numeric_df = df[iris.feature_names + ['target']]
corr_matrix = numeric_df.corr(numeric_only=True)
display(corr_matrix)

plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f')
plt.title('相関行列')
plt.show()
```
