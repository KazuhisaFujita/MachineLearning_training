### Step 0: 分析の流れを確認する
このノートブックでは、Iris データを使って可視化と相関の見方を練習します。各セルは上から順に実行し、図を見ながら特徴量の関係を確認してください.

```python
# ===== Step 0: ライブラリ読み込み =====
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

sns.set_theme(style='whitegrid')
```

### Step 1: データを読み込む
Iris データを DataFrame にして、後で可視化しやすい形に整えます。まずは先頭行を見て、列の構成を確認します。

```python
# ===== Step 1: データ読み込み =====
# TODO: IrisデータをDataFrameとして準備する
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

display(df.head())
```

### Step 2: pairplot を描く
特徴量どうしの関係をまとめて見て、品種ごとの差が出やすい場所を探します。色分けされた散布図の並び方に注目してください。

```python
# ===== Step 2: pairplot =====
# TODO: hue に使う列を確認する
pairplot_columns = iris.feature_names + ['species']
sns.pairplot(df[pairplot_columns], hue='species', corner=True)
plt.show()
```

### Step 3: 箱ひげ図でばらつきを見る
品種ごとの分布の違いを箱ひげ図で確認します。中央値やばらつきの差から、分類に効きそうな特徴量を探します。

```python
# ===== Step 3: 箱ひげ図 =====
# TODO: どの特徴量が品種ごとに分かれやすいか観察する
target_feature = 'petal length (cm)'  # TODO: 別の列にも変更して試す

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='species', y=target_feature)
plt.title(f'{target_feature} の箱ひげ図')
plt.show()
```

### Step 4: 相関行列を確認する
数値列どうしの関連の強さを相関行列で確認します。ヒートマップを見ると、どの特徴量が似た動きをするか分かります。

```python
# ===== Step 4: 相関行列 =====
# TODO: 数値列だけを使って相関行列を確認する
numeric_df = df[iris.feature_names + ['target']]
corr_matrix = numeric_df.corr(numeric_only=True)
display(corr_matrix)

plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f')
plt.title('相関行列')
plt.show()
```
