```python
# ===== Step 1: 変数 =====
# TODO: 数字を入れ替えて、結果がどう変わるか確認する
a = 10
b = 4

print('a =', a)
print('b =', b)
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)

# ヒント: = は値を変数に入れる記号です
```

```python
# ===== Step 2: ライブラリ読み込みとデータ準備 =====
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

plt.style.use('seaborn-v0_8')

# TODO: Irisデータセットを読み込む
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print('行数, 列数 =', df.shape)
display(df.head())

# ヒント: DataFrame は表のようにデータを扱うための型です
```

```python
# ===== Step 3: データ確認と可視化 =====
# TODO: 先頭5行、列名、基本統計量を確認する
print('列名一覧')
print(df.columns.tolist())
print('\n基本統計量')
display(df.describe())

print('\n品種ごとの件数')
display(df['species'].value_counts())

# TODO: ヒストグラムを描く
hist_column = 'sepal length (cm)'  # TODO: 別の列にも変えて試す
df[hist_column].plot(kind='hist', bins=15, figsize=(6, 4), title=f'{hist_column} のヒストグラム')
plt.xlabel(hist_column)
plt.show()

# TODO: 散布図を描く
x_col = 'sepal length (cm)'
y_col = 'petal length (cm)'

plt.figure(figsize=(6, 4))
for species_name, group in df.groupby('species'):
    plt.scatter(group[x_col], group[y_col], label=species_name, alpha=0.7)

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title('Irisの散布図')
plt.legend()
plt.show()

# ヒント: `df['列名']` で1つの列を取り出せます
```
