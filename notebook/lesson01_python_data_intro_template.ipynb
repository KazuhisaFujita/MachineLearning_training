## 加減乗除

まずは Python の簡単な計算が出来ることを確認します。セルを実行すると、計算式と結果が表示されます。

### 解説: Python の計算

Python では、数式に近い形で計算を書けます。

- `+`: 足し算
- `-`: 引き算
- `*`: 掛け算
- `/`: 割り算（結果は小数になります。例: `8 / 2` は `4.0`）
- `**`: べき乗（例: `2 ** 3` は $2^3 = 8$）

**ポイント:**

- 掛け算は `×` ではなく `*`、割り算は `÷` ではなく `/` を使います
- `//` を使うと「割り算の整数部分だけ」を取り出せます（例: `7 // 2` は `3`）
- `%` を使うと「割り算の余り」を取り出せます（例: `7 % 2` は `1`）
- このような計算に関わる記号を演算子といいます


````python
# ===== Step 0: 加減乗除 =====
# TODO: まずはセルを実行して、計算結果が下に表示されることを確認する
print('1 + 2 =', 1 + 2)
print('5 - 3 =', 5 - 3)
print('4 * 6 =', 4 * 6)
print('9 / 2 =', 9 / 2)
print('9 // 2 =', 9 // 2)  # 割り算の商（小数点以下を切り捨てる）
print('9 % 2 =', 9 % 2)  # 割り算の余り
print('2 ** 3 =', 2 ** 3)  # べき乗: 2の3乗

# シャープ（#）はコメントアウトの記号です。コードの実行には影響しません。
# +, -, *, / はそれぞれ足し算、引き算、掛け算、割り算です
# ** はべき乗（何乗するか）を表します
# print() は、() の中の値を出力する関数です。''の中の文字列はそのまま出力されます。,の後の値は計算されて出力されます。
````

## 変数を使って計算する

値を変数に入れてから、変数同士の計算を確認します。

### 解説: 変数とは

変数は、値に名前を付けて保存する箱のようなものです。一度変数に入れておけば、後でその名前を使って値を再利用できます。

``` python
a = 10
b = 4
print(a + b)  # 14
```

これは、次のように考えられます。

- `a = 10`: 「`a` という箱に `10` を入れる」という意味です
- `a + b`: 箱の中身同士を足します（`a`に10、`b`に4が入っているので、`10 + 4 = 14`になります）

変数を使うと、数字を変えたいときに数値の代入の箇所だけで済みます。つまり、数式は変更しなくてよいということです。

**ポイント:**

- `=` は「右の値を左の変数に入れる」操作です（数学の等号とは少し違います）。数学の「左と右が等しい」という意味ではないことに注意しましょう
- 変数名は半角英数字で、先頭は文字または `_` にします（例: `price`, `total_1`）。数字から始まる名前は使えません（例: `1price` はNG）
- 変数は何度でも上書きできます（例: `a = 10` の後で `a = 20` とすると、`a` は `20` になります）
- 変数名は「何が入っているか分かりやすい名前」をつけるのが良い慣習です（例: 面積なら `area`、横幅なら `width`）

#### 解説: 変数の型

Python では、値には**型**（データの種類）があります。同じ数字でも、整数と小数では型が違います。

| 型 | 例 | 説明 |
|---|---|---|
| `int` | `10`, `-3`, `0` | 整数 |
| `float` | `3.14`, `0.5`, `4.0` | 小数（浮動小数点） |
| `str` | `'hello'`, `'10'` | 文字列 |
| `bool` | `True`, `False` | 真偽（条件判断で使う） |

`type()` 関数で、変数に入っている値の型を確認できます。

- `type(10)` → `<class 'int'>`
- `type(3.14)` → `<class 'float'>`
- `type('10')` → `<class 'str'>`（`'10'` は文字列なので、数値の 10 とは別物です）

**ポイント:**

- `10`（int）と `10.0`（float）は数値としては同じですが、型は異なります
- `'10'`（str）は文字列なので、そのまま `10 + '10'` と足すことはできません（後述の型変換を参照）
- 型が混ざった計算（例: `10 + 2.5`）は、Python が自動的に `float` に変換して `12.5` を返します
- 条件分岐（`if`）で使う `True` / `False` は `bool` 型です

次のセルでは、`a` や `b` の値を変えて、計算結果がどう変わるか試してください。

````python
# ===== Step 1: 変数 =====
# TODO: 数字を入れ替えて、結果がどう変わるか確認する
a = 10
b = 4

print('a =', a)
print('b =', b)
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)
print('a // b =', a // b)
print('a % b =', a % b)

# 型を確認する: type() で値の種類を見られる
print('a の型 =', type(a))   # int（整数）
print('a / b の型 =', type(a / b))  # float（小数）

# ヒント: = は値を変数に入れる記号です
````

## 文字列とリスト

Pythonでは数値だけでなく、文字や複数の値も簡単に扱えます。ここでは「読みながら動かす」ことを重視してください。

### 解説: 文字列

文字列は、文字の並びのことを指します。文字列は `' '` や `" "` で囲んで表します。

- `'Iris'`: データセットの名前
- `'cm'`: 単位
- `'10'`: 数値 10 を文字列として表したもの

変数に文字列を入れることもできます（例: `name = 'Iris'`）。

**ポイント:**

- `'` と `"` はどちらでも使えますが、前後で揃える必要があります（例: `'Iris"` はエラーになります）
- 文字列同士は `+` で連結できます（例: `'Iris' + ' setosa'` → `'Iris setosa'`）

### 解説: 型変換（キャスト）

Python では、数値と文字列は別の型です。`'10'` は文字列なので、そのまま `10 + '10'` と足すことはできません。型を変えてから使うことを**型変換**（キャスト）と呼びます。

- `str(数値)`: 数値を文字列に変換（例: `str(75)` → `'75'`）
- `int(文字列)`: 文字列を整数に変換（例: `int('42')` → `42`）
- `float(文字列)`: 文字列を小数に変換（例: `float('3.14')` → `3.14`）

**ポイント:**

- 文字列と数値を `+` で連結したいときは、先に `str()` で数値を文字列に変換します（例: `str(75) + '点'` → `'75点'`）
- 逆に、文字列から数値を取り出して計算したいときは `int()` や `float()` を使います
- 変換できない値（例: `int('abc')`）はエラーになります

### 解説: リスト

リストは、複数の値を順番に並べたものです。値は何でもよいです。角括弧 `[ ]` で囲み、要素を `,` で区切ります。
- `values = [1, 2, 3, 4, 5]`
- `features = ['sepal length', 'sepal width', 'petal length', 'petal width']`
- `len(features)`: リストの長さ（要素数）を確認
- `features[0]`: 1つ目の要素にアクセス（Python では番号は0から始まります）
- `features[-1]`: 末尾の要素にアクセス

**ポイント:**

- リストの番号は0から始まるので、`features[0]` が最初、`features[1]` が2番目になります。これは「先頭からの距離」と考えると分かりやすいです（先頭は0番目、1つ先は1番目）
- `-1` を使うと最後の要素、`-2` で後ろから2番目にアクセスできます（例: 4つの要素がある場合、`features[-1]` は4つ目、`features[-2]` は3つ目）
- `append()` でリストの最後に新しい要素を追加できます（例: `features.append('species')` で `'species'` が5つ目の要素として追加されます）
- リストには文字列だけでなく、数値も混ぜて入れることができます（例: `[1, 'two', 3]`）

次のセルで文字列とリストを試してください。

````python
# ===== Step 2-1: 文字列とリスト =====
name = 'Iris'
unit = 'cm'

print('データ名:', name)
print('単位:', unit)

# 型変換の例: 数値を文字列に変換して連結する
n = 150
print('データ件数:', str(n) + '件')  # str() で数値を文字列にしてから + で連結。こうすることで、文字列と数値を一緒に表示できる

features = ['sepal length', 'sepal width', 'petal length', 'petal width']
print('特徴量一覧:', features)
print('特徴量の数:', len(features))
print('最初の特徴量:', features[0])

# インデックスを -1 にすると、最後の要素を取り出せる
print('最後の特徴量:', features[-1])

# append() で末尾に要素を追加する
features.append('species')
print('追加後の特徴量一覧:', features)
````

## 制御構文（条件分岐・繰り返し）

制御構文とは、プログラムの実行の流れを制御するための構文で、代表的なものに条件分岐を表すif文や、繰り返しを表すfor文があります。

### 解説: 条件分岐 (`if`)

条件によって処理を分けるときに `if` を使います。

- `score >= threshold`: 「score が threshold 以上か」を調べます
- 条件が真（True）なら `if` の中のブロックが実行されます
- 偽（False）なら `else` の中のブロックが実行されます

**ポイント:**

- `if` や `else` の行の最後には `:` （コロン）が必要です。これを忘れると `SyntaxError` になります
- ブロックの中は字下げ（インデント）して書きます。Python では、このインデントで「どこからどこまでが `if` の中か」を判断します。通常は半角スペース4個を使います
- 比較には次の記号を使います: `>=`（以上）、`<=`（以下）、`>`（より大きい）、`<`（より小さい）、`==`（等しい）、`!=`（等しくない）
- `==` と `=` は違います。`=` は代入（変数に値を入れる）、`==` は比較（等しいかどうか調べる）です

次のセルで条件分岐を試してください。

````python
# ===== Step 2-2: 条件分岐 =====
score = 75
threshold = 60

print('score =', score)
print('threshold =', threshold)

if score >= threshold:
    print(score, 'は合否ライン', threshold, '以上です')
else:
    print(score, 'は合否ライン', threshold, '未満です')

# TODO: score または threshold を変えて、表示がどう変わるか確認する
````

### 繰り返し (`for`)

繰り返し処理には`for`を使います。繰り返し処理とは、例えば、リストの要素を1つずつ処理したり、表示したり、するときに使えます。

- `for fruit in fruits:`: リスト `fruits` の要素を順番に `fruit` に入れて処理します
- `enumerate(fruits)`: 番号（index）と値の両方を同時に取得できます。`for index, fruit in enumerate(fruits):` と書くと、`index` に0からの番号、`fruit` に値が入ります
- `range(n)`: 0 から n-1 までの整数を順番に生成します。`range(start, stop, step)` で開始値・終了値・ステップ幅を指定できます

**ポイント:**

- `for` の行の最後には `:` （コロン）が必要です。`if` と同じです
- ブロックの中身はインデントして書きます
- ループは、リストの要素がなくなるまで自動的に繰り返されます
- ブロックの中身は、各要素に対して1回ずつ実行されます
- `range` を使うと、特定の回数だけループを回すことができます


````python
# ===== Step 2-3: for ループ（基本）=====
# リストを使った for ループの例です。リストの中身を順番に取り出して、print() で表示します。
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(fruit)

# TODO: fruits の中身を変えて、ループの動きを確認する
````

````python
# ===== Step 2-4: for ループ =====
# enumerate() を使った for ループの例です。リストの中身を順番に取り出して、インデックス番号と一緒に print() で表示します。
fruits = ['apple', 'banana', 'cherry']

for index, fruit in enumerate(fruits):
    print(index, ':', fruit)

# TODO: fruits の中身を変えて、ループの動きを確認する
````

````python
# ===== Step 2-5: range を使った for ループ =====
# range(n): 0 から n-1 までの整数を生成
print("--- range(5) ---")
for i in range(5):
    print(i, end=' ')  # end=' ' は改行の代わりに空白を1つ入れる指定です
print()  # 最後に改行だけ出力する

# range(start, stop): start から stop-1 まで
print("--- range(1, 6) ---")
for i in range(1, 6):
    print(i, end=' ')
print()

# range(start, stop, step): step ずつ増やす
print("--- range(0, 10, 2) ---")
for i in range(0, 10, 2):
    print(i, end=' ')
print()

# TODO: range の引数を変えて、出力がどう変わるか確認する
# ヒント: print(i, end=' ') の end=' ' を end=', ' に変えると、区切り文字が変わります
````

## 関数を作る

同じ計算を何度も使いたいときは、関数にまとめると便利です。まずは数学の関数 f(x) と同じだと思ってください。

関数では、
- 入力: 関数に渡す値（引数と呼びます）
- 出力: 関数が返す値（戻り値と呼びます）

まずは長方形の面積を求める関数を読んで、入力を変えて試してください。

````python
# ===== Step 3: 関数 =====
# def で関数を定義します。関数名(引数1, 引数2, ...) と書きます
# return で結果（戻り値）を返します
def rectangle_area(width, height):
    return width * height

print('3 x 5 の面積 =', rectangle_area(3, 5))
print('7 x 2 の面積 =', rectangle_area(7, 2))

# TODO: 引数を変えて、面積がどう変わるか確認する
# ヒント: rectangle_area(10, 20) のように、好きな数字を入れてみてください
````

このように、面積を求める計算式を関数にまとめられます。同じ計算を何度も書く必要がなくなり、式を直すときも関数の中だけを変更すればよいため、間違いを減らして管理しやすくなります。
