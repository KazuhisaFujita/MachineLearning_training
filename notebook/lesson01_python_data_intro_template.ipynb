#### 解説: Python の計算

$$
面積 = 横幅 \times 高さ
$$

コードでは `width * height` のように書きます。


````python
# ===== Step 0: 加減乗除 =====
# TODO: まずはセルを実行して、計算結果が下に表示されることを確認する
print('1 + 2 =', 1 + 2)
print('5 - 3 =', 5 - 3)
print('4 * 6 =', 4 * 6)
print('8 / 2 =', 8 / 2)

# ヒント: +, -, *, / はそれぞれ足し算、引き算、掛け算、割り算です
````

````python
# ===== 追加演習: 面積を計算する =====
# TODO: width と height の値を変えて、面積がどう変わるか確認する
width = 3
height = 5
area = width * height

print('横幅 =', width)
print('高さ =', height)
print('面積 =', area)
````

### Step 3: 関数を作る

同じ計算を何度も使いたいときは、関数にまとめると便利です。

関数では、
- 入力: 関数に渡す値
- 出力: 関数が返す値

まずは長方形の面積を求める関数を読んで、入力を変えて試してください。

````python
# ===== Step 3: 関数 =====
def rectangle_area(width, height):
    return width * height

print('3 x 5 の面積 =', rectangle_area(3, 5))
print('7 x 2 の面積 =', rectangle_area(7, 2))

# TODO: 引数を変えて、面積がどう変わるか確認する
````

### Step 4: NumPy で配列と行列を扱う

データ分析では、数値が並んだ配列をまとめて扱うことが多くあります。

`NumPy` は、そのための基本ライブラリです。ここでは 2 次元配列を「行と列を持つ表に近いもの」として見ます。

この段階では、
- 形を確認する
- 行ごと、列ごとに平均を出す

ところまでできれば十分です。

````python
# ===== Step 4: NumPy 配列 =====
import numpy as np

scores = np.array([
    [80, 65, 70],
    [90, 88, 92],
    [75, 60, 78]
])

print('scores =')
print(scores)
print('形 =', scores.shape)
print('各行の平均 =', scores.mean(axis=1))
print('各列の平均 =', scores.mean(axis=0))

# TODO: 1つ値を変えて、平均がどう変わるか確認する
````

### Step 5: ライブラリを読み込んでデータを準備する

ここからは、次回以降の分析で使う表形式データに少し触れます。

ただし、この回の中心は Python の基本操作です。グラフはまだ描きません。

`pandas` は表を扱うためのライブラリ、`load_iris` は練習用データを読み込むための関数です。

#### 解説: 表形式データと DataFrame

Iris データセットは、花の特徴を行と列で表した表形式データです。

- 1行: 1つの花の観測データ
- 1列: 花びらの長さ、がく片の幅などの特徴量
- `species`: 花の種類

データ分析では、まず次を確認します。

- 行数と列数
- 列名
- 先頭数行
- 値の範囲や平均

これは「モデルを作る前に、データが何を表しているかを理解する」ためです。


````python
# ===== Step 5: ライブラリ読み込みとデータ準備 =====
import pandas as pd
from sklearn.datasets import load_iris

# TODO: Irisデータセットを読み込む
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print('行数, 列数 =', df.shape)
display(df.head())

# ヒント: DataFrame は表のようにデータを扱うための型です
````

### Step 6: データを確認する

列名、基本統計量、件数を確認します。

この回では「表として何が入っているか」を確認するところまでに留めます。

分布や関係をグラフで見る作業はテーマ2で行います。

````python
# ===== Step 6: データ確認 =====
# TODO: 先頭5行、列名、基本統計量を確認する
print('列名一覧')
print(df.columns.tolist())
print('\n基本統計量')
display(df.describe())

print('\n品種ごとの件数')
display(df['species'].value_counts())

# ヒント: `df['列名']` で1つの列を取り出せます
#       df.describe() で各列の平均、標準偏差、最小値、最大値などが確認できます
````
