### Step 0: 加減乗除の確認

#### 解説: Python の計算

Python では、数式に近い形で計算を書けます。

- `+`: 足し算
- `-`: 引き算
- `*`: 掛け算
- `/`: 割り算
- `**`: べき乗

例えば、長方形の面積は次のように考えられます。

$$
面積 = 横幅 	imes 高さ
$$

コードでは `width * height` のように書きます。


```python
# ===== Step 0: 加減乗除 =====
# TODO: まずはセルを実行して、計算結果が下に表示されることを確認する
print('1 + 2 =', 1 + 2)
print('5 - 3 =', 5 - 3)
print('4 * 6 =', 4 * 6)
print('8 / 2 =', 8 / 2)

# ヒント: +, -, *, / はそれぞれ足し算、引き算、掛け算、割り算です
```

```python
# ===== 追加演習: 面積を計算する =====
# TODO: width と height の値を変えて、面積がどう変わるか確認する
width = 3
height = 5
area = width * height

print('横幅 =', width)
print('高さ =', height)
print('面積 =', area)
```

### Step 2: ライブラリを読み込んでデータを準備する
Pandas と Matplotlib を使う準備をして、Iris データセットを DataFrame にまとめます。ここでは表形式のデータとして扱える形に変換します.

#### 解説: 表形式データと DataFrame

Iris データセットは、花の特徴を行と列で表した表形式データです。

- 1行: 1つの花の観測データ
- 1列: 花びらの長さ、がく片の幅などの特徴量
- `species`: 花の種類

データ分析では、まず次を確認します。

- 行数と列数
- 列名
- 先頭数行
- 値の範囲や平均

これは「モデルを作る前に、データが何を表しているかを理解する」ためです。


```python
# ===== Step 2: ライブラリ読み込みとデータ準備 =====
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Hiragino Sans', 'Yu Gothic', 'Arial Unicode MS', 'DejaVu Sans']

plt.style.use('seaborn-v0_8')

# TODO: Irisデータセットを読み込む
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print('行数, 列数 =', df.shape)
display(df.head())

# ヒント: DataFrame は表のようにデータを扱うための型です
```

### Step 3: データを確認して可視化する
列名、基本統計量、件数を確認したあと、ヒストグラムと散布図でデータの分布を見ます。列を変えると、見え方の違いも確認できます.

```python
# ===== Step 3: データ確認と可視化 =====
# TODO: 先頭5行、列名、基本統計量を確認する
print('列名一覧')
print(df.columns.tolist())
print('\n基本統計量')
display(df.describe())

print('\n品種ごとの件数')
display(df['species'].value_counts())

# TODO: ヒストグラムを描く
hist_column = 'sepal length (cm)'  # TODO: 別の列にも変えて試す
df[hist_column].plot(kind='hist', bins=15, figsize=(6, 4), title=f'{hist_column} のヒストグラム')
plt.xlabel(hist_column)
plt.show()

# TODO: 散布図を描く
x_col = 'sepal length (cm)'
y_col = 'petal length (cm)'

plt.figure(figsize=(6, 4))
for species_name, group in df.groupby('species'):
    plt.scatter(group[x_col], group[y_col], label=species_name, alpha=0.7)

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title('Irisの散布図')
plt.legend()
plt.show()

# ヒント: `df['列名']` で1つの列を取り出せます
```
