### Step 1: 変数を使って計算する

#### 解説: 変数とは

変数は、値に名前を付けて保存する箱のようなものです。一度変数に入れておけば、後でその名前を使って値を再利用できます。

$$
a = 10,\quad b = 4 \implies a + b = 14
$$

例えば、次のように考えられます。

- `a = 10`: 「`a` という箱に `10` を入れる」という意味です
- `a + b`: 箱の中身同士を足します（`10 + 4 = 14`）

変数を使うと、数字を変えたいときに1箇所だけ直せば良くなります。

**ポイント:**

- `=` は「右の値を左の変数に入れる」操作です（数学の等号とは少し違います）。数学の「左と右が等しい」という意味ではないことに注意しましょう
- 変数名は半角英数字で、先頭は文字または `_` にします（例: `price`, `total_1`）。数字から始まる名前は使えません（例: `1price` はNG）
- 変数は何度でも上書きできます（例: `a = 10` の後で `a = 20` とすると、`a` は `20` になります）
- 変数名は「何が入っているか分かりやすい名前」をつけるのが良い慣習です（例: 面積なら `area`、横幅なら `width`）

次のセルでは、`a` や `b` の値を変えて、計算結果がどう変わるか試してください。

````python
# ===== Step 1: 変数 =====
# TODO: 数字を入れ替えて、結果がどう変わるか確認する
a = 10
b = 4

print('a =', a)
print('b =', b)
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)

# ヒント: = は値を変数に入れる記号です
````

### Step 2: 文字列・リスト・条件分岐・繰り返し

数値の計算だけでなく、文字や複数の値も扱ってみます。ここでは「読みながら動かす」ことを重視してください。

#### 解説: 文字列

文字列は、文字の並びを `' '` や `" "` で囲んで表します。

- `'Iris'`: データセットの名前
- `'cm'`: 単位

変数に文字列を入れることもできます（例: `name = 'Iris'`）。

**ポイント:**

- 文字列同士は `+` で連結できます（例: `'Iris' + ' setosa'` → `'Iris setosa'`）

#### 解説: リスト

リストは、複数の値を順番に並べたものです。角括弧 `[ ]` で囲み、要素を `,` で区切ります。

- `features = ['sepal length', 'sepal width', 'petal length', 'petal width']`
- `len(features)`: リストの長さ（要素数）を確認
- `features[0]`: 1つ目の要素にアクセス（Python では番号は0から始まります）
- `features[-1]`: 最後尾の要素にアクセス

**ポイント:**

- リストの番号は0から始まるので、`features[0]` が最初、`features[1]` が2番目になります。これは「先頭からの距離」と考えると分かりやすいです（先頭は0番目、1つ先は1番目）
- `-1` を使うと最後の要素、`-2` で後ろから2番目にアクセスできます（例: 4つの要素がある場合、`features[-1]` は4つ目、`features[-2]` は3つ目）
- `append()` でリストの最後に新しい要素を追加できます（例: `features.append('species')` で `'species'` が5つ目の要素として追加されます）
- リストには文字列だけでなく、数値も混ぜて入れることができます（例: `[1, 'two', 3]`）

#### 解説: 条件分岐 (`if`)

条件によって処理を分けるときに `if` を使います。

- `score >= threshold`: 「score が threshold 以上か」を調べます
- 条件が真（True）なら `if` の中のブロックが実行されます
- 偽（False）なら `else` の中のブロックが実行されます

**ポイント:**

- `if` や `else` の行の最後には `:` （コロン）が必要です。これを忘れると `SyntaxError` になります
- ブロックの中は字下げ（インデント）して書きます。Python では、このインデントで「どこからどこまでが `if` の中か」を判断します。通常は半角スペース4個を使います
- 比較には次の記号を使います: `>=`（以上）、`<=`（以下）、`>`（より大きい）、`<`（より小さい）、`==`（等しい）、`!=`（等しくない）
- `==` と `=` は違います。`=` は代入（変数に値を入れる）、`==` は比較（等しいかどうか調べる）です

次のセルで条件分岐を試してください。

````python
# ===== Step 2-2: 条件分岐 =====
score = 75
threshold = 60

print('score =', score)
print('threshold =', threshold)

if score >= threshold:
    print(score, 'は合否ライン', threshold, '以上です')
else:
    print(score, 'は合否ライン', threshold, '未満です')

# TODO: score または threshold を変えて、表示がどう変わるか確認する
````

#### 解説: 繰り返し (`for`)

リストの要素を1つずつ処理するときに `for` を使います。

- `for fruit in fruits:`: リスト `fruits` の要素を順番に `fruit` に入れて処理します
- `enumerate(fruits)`: 番号（index）と値の両方を同時に取得できます。`for index, fruit in enumerate(fruits):` と書くと、`index` に0からの番号、`fruit` に値が入ります
- `range(n)`: 0 から n-1 までの整数を順番に生成します。`range(start, stop, step)` で開始値・終了値・ステップ幅を指定できます

**ポイント:**

- `for` の行の最後には `:` （コロン）が必要です。`if` と同じです
- ブロックの中身はインデントして書きます
- ループは、リストの要素がなくなるまで自動的に繰り返されます
- ブロックの中身は、各要素に対して1回ずつ実行されます
- `range` を使うと、特定の回数だけループを回すことができます


````python
# ===== Step 2-2: for ループ（基本）=====
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(fruit)

# TODO: fruits の中身を変えて、ループの動きを確認する
````

````python
# ===== Step 2-3: for ループ =====
fruits = ['apple', 'banana', 'cherry']

for index, fruit in enumerate(fruits):
    print(index, ':', fruit)

# TODO: fruits の中身を変えて、ループの動きを確認する
````

````python
# ===== Step 2-4: range を使った for ループ =====
# range(n): 0 から n-1 までの整数を生成
print("--- range(5) ---")
for i in range(5):
    print(i, end=' ')  # end=' ' は改行の代わりに空白を1つ入れる指定です
print()  # 最後に改行だけ出力する

# range(start, stop): start から stop-1 まで
print("--- range(1, 6) ---")
for i in range(1, 6):
    print(i, end=' ')
print()

# range(start, stop, step): step ずつ増やす
print("--- range(0, 10, 2) ---")
for i in range(0, 10, 2):
    print(i, end=' ')
print()

# TODO: range の引数を変えて、出力がどう変わるか確認する
# ヒント: print(i, end=' ') の end=' ' を end=', ' に変えると、区切り文字が変わります
````


**発展: データ分析では for ループを避ける理由**

データ分析では、`for` ループで1つずつ計算すると非常に遅くなります。その代わりに、NumPy や pandas の「ベクトル化」機能を使って、配列全体を一度に処理します。

例えば、100万個の値にそれぞれ 2 を掛ける場合:

- `for` ループ: 約 0.1 秒

### Step 3: 関数を作る

### Step 4: NumPy で配列と行列を扱う

データ分析では、数値が並んだ配列をまとめて扱うことが多くあります。

`NumPy` は、数値計算を高速に行うための基本ライブラリです。Python のリストと比べて、次のような特徴があります。

- **高速**: C 言語で実装されており、大量の数値を一度に処理できます
- **簡潔**: 配列全体に対する計算が、1 行のコードで書けます
- **行列計算**: 2 次元配列を「行列」として扱い、行列演算が容易です

ここでは 2 次元配列を「ベクトルの集まり、つまり行列」として見ます。さらに高次元配列は、「行列の集まり」として捉えることができます。

この段階では、次の操作を確認できれば十分です。

- 配列の形（shape）を確認する
- 行ごと、列ごとに平均を計算する
- 配列同士の四則演算を行う

平均以外にも、積和・内積・転置など、様々な行列演算の機能を使うことができます。

**用語の確認:**

- **配列（array）**: NumPy で数値を並べたもの。Python のリストに似ていますが、数値計算に特化しています
- **shape**: 配列の形を表します。2次元配列の場合は `(行数, 列数)` で表示されます
- **axis**: 平均などを計算するときの「方向」を指定します。`axis=0` は列方向（各列の平均）、`axis=1` は行方向（各行の平均）です

````python
# ===== Step 4: NumPy 配列 =====
import numpy as np  # NumPy を読み込み、np という短い名前で使えるようにする

scores = np.array([
    [80, 65, 70],
    [90, 88, 92],
    [75, 60, 78]
])

print('scores =')
print(scores)
print('形 =', scores.shape)  # (行数, 列数) = (3, 3)
print('各行の平均 =', scores.mean(axis=1))  # 行方向（横）の平均
print('各列の平均 =', scores.mean(axis=0))  # 列方向（縦）の平均

# NumPy の強み: 配列全体を一度に計算できる
print('全要素に 10 点加算 =')
print(scores + 10)
print('全要素を 2 倍 =')
print(scores * 2)

# TODO: 1つ値を変えて、平均がどう変わるか確認する
# TODO: scores + 10 の結果が、元の scores を書き換えないことを確認する
#       （再度 print(scores) を実行して、元の値のままであることを確かめてください）
````

````python
# ===== Step 5: ライブラリ読み込みとデータ準備 =====
import pandas as pd
from sklearn.datasets import load_iris

# scikit-learn に同梱されている Iris データセットを読み込む
iris = load_iris(as_frame=True)

# DataFrame としてコピーする
df = iris.frame.copy()

# target (数値) を species (品種名) に変換する
# target は 0, 1, 2 の数値で品種を表しています。これを 'setosa' などの名前に変換します
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print('行数, 列数 =', df.shape)
display(df.head())  # 先頭5行を表形式で表示する

# ヒント: DataFrame は表のようにデータを扱うための型です
#       load_iris() は scikit-learn に同梱された練習用データです
#       display() は Jupyter で表をきれいに表示する関数です（print() より見やすい形式になります）
````
